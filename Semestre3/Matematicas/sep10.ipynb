```python
import numpy as np 
import matplotlib.pyplot as plt
import scipy.stats as st
```

```python
def cdf(z1:float=0):
    dist = st.norm(loc=0, scale=1)
    z1_arary = np.arange(-4,z1,0.01)
    p_z1= dist.pdf(z1_arary)
    z= np.arange(-4,4,0.01)
    P_z = dist.pdf(z)
    plt.plot(z,P_z)
    plt.axvline(x = z1, color='red', ymax=dist.pdf(z1)*2.5)
    plt.fill_between(z1_arary, p_z1)
    plt.ylim(0,0.40)
    plt.xlim(-4,4)
    plt.title('Distribucion normal: Area cdf')
    plt.show()
    area= dist.cdf(z1)
    print(f'area:{round(area, 4)} ')


#Area bajo la curva
```

```python

cdf(5)
```

```python
def cdf_c(z1:float=0):
    dist = st.norm(loc=0, scale=1)
    z1_arary = np.arange(z1,4,0.01)
    p_z1= dist.pdf(z1_arary)
    z = np.arange(-4,4,0.01)
    P_z = dist.pdf(z)
    plt.plot(z,P_z)
    plt.axvline(x = z1, color='red', ymax=dist.pdf(z1)*2.5)
    plt.fill_between(z1_arary, p_z1)
    plt.ylim(0,0.40)
    plt.xlim(-4,4)
    plt.title('Distribucion normal: Area cdf')
    plt.show()
    area= 1 -dist.cdf(z1)
    print(f'area:{round(area, 4)} ')


#Area bajo la curva
```

```python
cdf_c(1.5)
```

```python

def cdf_bet(z1:float=-1, z2:float = 1):
    dist = st.norm(loc=0, scale=1)
    zt_arary = np.arange(z1,z2,0.01)
    p_z1= dist.pdf(zt_arary)
    

    z = np.arange(-4,4,0.01)
    P_z = dist.pdf(z)
    plt.plot(z,P_z)
    plt.axvline(x = z1, color='red', ymax=dist.pdf(z1)*2.5)
    plt.axvline(x = z2, color='red', ymax=dist.pdf(z2)*2.5)
    plt.fill_between(zt_arary, p_z1)
    plt.ylim(0,0.40)
    plt.xlim(-4,4)
    plt.title('Distribucion normal: Area cdf')
    plt.show()
    area=  dist.cdf(z2)- dist.cdf(z1)
    print(f'area:{round(area, 4)} ')


#Area bajo la curva
```

```python
cdf_bet(-2,2)
```
