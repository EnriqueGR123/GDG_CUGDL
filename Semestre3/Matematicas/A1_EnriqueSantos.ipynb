```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
#var, std, max, min, q1, q_i3
print(df['Price'].var(ddof=0))
print(f'Desviacion estandar {df['Price'].std()}')
print(f'Maximo: {df['Price']. max()}')
print(f'Minimo: {df['Price'].min()}')
print(f'Cuartil 1: {df['Price'].quantile(0.25)}')
print(f'Cuartil 3 {df['Price'].quantile(0.75)}')
```

## MEDIA, MEDIANA, MODA, VAR, STD, MAX, MIN, Q1, Q3 De cada variablw

# 9) Realiza diagramas de dispersión entre la variable precio y las variables pies cuadrados construidos, número de cuartos, número de baños, y número de ofertas. ¿Qué se puede concluir de los gráficos?


```python
# Price	SqFt	Bedrooms	Bathrooms	Offers	Brick	Neighborhood
#Price vs. SqFt Price vs. Offers 
cols = ['SqFt', 'Bedrooms', 'Bathrooms', 'Offers']

for col in cols:
    plt.figure(figsize=(6,3))
    plt.scatter(x=df['Price'], y=df[col])
    plt.title(f'Price and {col} scatter')
    plt.ylabel('Price')
    plt.xlabel(col)
    plt.show()
```

# Conclusiones 
### Gráfica 1: Price vs. SqFt 
- Hay una relación positiva clara. Entre más grandes son los pies cuadrados (SqFt), más cara es la casa (Price). Los puntos suben en diagonal hacia la derecha.2. 

### Gráfica 2: Price vs. Bedrooms
- La mayoría de tus casas tienen 3 habitaciones. Las casas con 2 habitaciones son notablemente más baratas (se quedan abajo en el precio), mientras que las de 4 y 5 habitaciones logran alcanzar los precios más altos del mercado.3. 

### Gráfica 3: Price vs. Bathrooms
- Tener más baños sube el valor. Las casas con 2 baños tienen un límite de precio más bajo, las de 3 baños dominan el rango medio-alto, y la casa con 4 baños es una de las más caras de todo tu registro.4. 

### Gráfica 4: Price vs. Offers
- Aquí la relación es inversa. Las casas que reciben pocas ofertas (1 o 2) suelen ser las que se venden más caro. Esto pasa normalmente porque las casas de lujo o muy caras tienen menos compradores interesados y tardan más en recibir ofertas, a diferencia de las casas baratas que se llenan de ofertas rápido.
