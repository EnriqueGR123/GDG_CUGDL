### Ejercicio 2

- Un ingeniero quimico afirma que el rendimiento medio de loa poblacion de cierto proceso en lotes es 500gr c/una de materia prima. 
Para verifica dicha afirmacion muestre 25 lotes de cada uno. Si el valor calculado de t cae entre -t0.05 y t0.05, queda satisfecho con su afirmacion. ¿Que conclusion deberia obtener una muestra que tiene x- 518 gr/m y desv estandar muestral de S=40 gr/mno Suponga que la distribucion de rendimiento es aproximadamente gaussiana v=n-1 n=25 ta=t0.05 X = 518 s=40

## $$
h(t;\nu) = \frac{\Gamma[(\nu +1)/2]} {\Gamma(\nu/2)\sqrt{\pi\nu}}\left(1+\frac{t^2}{\nu}\right)^{-(\nu+1)/2}
$$

```python
def tdist(t,nu):
    return  gamma((nu+1)*0.5 )/(gamma(nu*0.5)*np.sqrt(np.pi*nu)) * (1.0 +t**2/nu) **(-(nu+1.0)*0.5)
```

```python
nu = 14
x_array = np.linspace(-5,5, 1000)
y_array= tdist(x_array, nu)
plt.plot(x_array, y_array, linestyle ='-', color='blue')
plt.show()
```

```python
integrate.quad(tdist, -np.inf, np.inf, args=(nu))
```

## Ejercicio 1

```python
nu=14
t_arr = np.linspace(-10,10,1000)

ss = 0
for tval in t_arr:
    prob = integrate.quad(tdist, -np.inf, tval, args=(nu))
    if prob[0] >= 0.025:
        print(t_arr[ss-1])
        break
    ss+=1
```
