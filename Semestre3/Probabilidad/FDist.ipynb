```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate
from scipy.special import gamma
```

## $$h(f;\,\nu_1,\nu_2)=
\begin{cases}
\frac{\Gamma[(\nu_1+\nu_2)/2](\nu_1/\nu_2)^{(\nu_1/2)}}
{\Gamma(\nu_1/2)\Gamma(\nu_2/2)}
\frac{f^{(\nu_1/2-1)}}
{(1+\nu_1 f/\nu_2)^{(\nu_1+\nu_2)/2}},
& f>0,\\
0, & f\leq 0
\end{cases}$$


```python
def fdist(f,nu1,nu2):
    return gamma((nu1+nu2)*0.5) * (nu1/nu2)**(nu1*0.5)/(gamma(nu1*0.5) * gamma(nu2*0.5)) *f**(nu1*0.5 -1.0)/(1.0+nu1*f/nu2) **((nu1+nu2) *0.5)
```

```python
nu1= 6.0
nu2=12.0
x_array = np.linspace(0.000001, 6, 1000)
y_array = fdist(x_array, nu1,nu2)
plt.plot(x_array, y_array, linestyle='-', color='blue', label=r'$\nu_1 = %.1f, \nu_2 = %.1f$'%(nu1, nu2))

plt.xlabel(r'$f$', fontsize=16)
plt.ylabel(r'$F(f;\,\nu_1,\,\nu_2)$', fontsize=16)
plt.show()
```

```python
integrate.quad(fdist, 0.000001, np.inf, args=(nu1,nu2))
```
