## $$ f(x; \nu) = \frac{1}{2^{\nu/2}\Gamma(\nu/2)} x^{\nu/2-1} e^{-x/2}, \; x>0 $$ 

```python
def chi(x,nu):
    return 1.0/(2**(nu*0.5)*scipy.special.gamma(nu*0.5))*x**(nu*0.5 -1.0)*np.exp(-x*0.5)
```
