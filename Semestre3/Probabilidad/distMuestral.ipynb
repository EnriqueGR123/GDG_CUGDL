```python
#ARRAY DE PROMEDIOS 
muestra = 500
mu = 0.0 #Media poblacional 
sigma = 1.0 #Varianza
data_syn = np.random.normal(mu, sigma, muestra)
mean_data_syn = np.mean(data_syn) #Media
```

```python
#Array de promedio de los promedios
nexp = 1000
mean_exp = np.zeros(nexp, dtype=np.float64)
for iexp in range(nexp):
    data_syn = np.random.normal(mu, sigma, muestra)
    mean_exp[iexp] = np.mean(data_syn)*np.sqrt(muestra)
    
```

```python
def gauss(x, mu, sigma):
    return 1.0/(np.sqrt(2.0*np.pi)*sigma)*np.exp(-(x - mu)**2/(2.0*sigma**2))
```

```python
#mean_exp
```

```python
plt.hist(mean_exp, bins=30, density=True, histtype='step', color='red')
x_array = np.linspace(-10,10,1000)
gaus_dist = gauss(x_array, mu, sigma)
plt.plot(x_array, gaus_dist)
plt.show()
```
