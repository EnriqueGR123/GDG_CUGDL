# Propón un ejemplo de problema supervisado/no supervisado  que pueda resolverse con estos datos.
- Supervisado: Se usa cuando queremos predecir un dato específico apoyándonos en el histórico de la tabla. Al tener etiquetas claras (como el precio o el vecindario), el modelo aprende de los errores comparando sus predicciones con la realidad.
- No Supervisado: Se usa cuando no buscamos predecir nada en particular, sino encontrar patrones ocultos o descubrir cómo se agrupan los datos de forma natural sin intervención humana.
- Algun ejemplo podria ser crear un cotizador automático para una inmobiliaria, con los datos se podrian agruparan las casas, encasasa pequeñas, grandes, de ladrillo etc, esto nos podira facilitar y ayudar a entender los tipos de clientes sin etiquetarlos manuealmente
